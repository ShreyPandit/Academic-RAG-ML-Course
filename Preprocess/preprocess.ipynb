```python
import nltk
import os
import json
```

```python
all_data = {}
```

```python
def preprocess_transcript(data, search_for=None):
    extracted_data = []
    for line in data:
        if search_for is not None:
            if line.startswith(search_for):
                extract = ' '.join(line.split(':')[1:]).replace('\n', '')
                if extract:
                    extracted_data.append(extract)
        else:
            extract = line.replace('\n', '')
            if extract:
                extracted_data.append(extract)
    print(f"Extracted {len(extracted_data)} lines from a total of {len(data)} lines.")
    return extracted_data
```

```python
## Extract merged_transcript.txt
```

```python
f = open('/var/local/adityav/Projects/temp/mlwb/merged_transcript.txt', 'r')
data = []
for line in f:
    data.append(line)
f.close()
print(f"Read {len(data)} lines from merged_transcript.txt")
```

```python
extracted_data = preprocess_transcript(data, "Inderjit Dhillon:")
```

```python
print(data[:10])
print(extracted_data[:10])
```

```python
all_data['merged_transcript.txt'] = extracted_data
```

```python
## Extract Transcript_Feb19_Mar5.txt
```

```python
f = open('/var/local/adityav/Projects/temp/mlwb/Transcript_Feb19_Mar5.txt', 'r')
data = []
for line in f:
    data.append(line)
f.close()
print(f"Read {len(data)} lines from Transcript_Feb19_Mar5.txt")
```

```python
## Extract Books
```

```python
books_list = [i for i in os.listdir('/var/local/adityav/Projects/temp/mlwb/books_txt/') if i.endswith('.txt')]
print(books_list)
```

```python
for book_name in books_list:
    f = open(f'/var/local/adityav/Projects/temp/mlwb/books_txt/{book_name}', 'r')
    data = []
    for line in f:
        data.append(line)
    f.close()
    print(f"Read {len(data)} lines from {book_name}")
    all_data[book_name] = preprocess_transcript(data)
```
